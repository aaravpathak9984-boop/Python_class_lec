## **Tasks on Data Set**

```python
## calculate total no. of loan applications..

no_of_app=df['id'].nunique();   ## to give unique value

print(f"Total applications: {no_of_app}")


## current month (june) applications..
june_app = df[df['issue_date'].dt.month==6]

no_of_june_app = june_app['id'].nunique()

print(f"June applications: {no_of_june_app}")
```

```python
current_date =pd.Timestamp.today() ## to get current date and time

# cd=pd.Timestamp.today()

cur_month=current_date.month

cur_date=current_date.day

app_in_mon=df[df['issue_date'].dt.month==cur_month]
no_of_mnth_app=app_in_mon['id'].nunique()


# Fixed condition
app_till_cd = df[(df['issue_date'].dt.month == cur_month) & (df['issue_date'].dt.day <= cur_date)]

no_of_app=app_till_cd['id'].nunique()


print(f"Current month applications: {no_of_mnth_app} and curent day applications are : {no_of_app}")
```

```python
total_loan_in_month_date=app_till_cd['loan_amount'].sum()

print(f"Total amonut invested in this month till date is : {round(total_loan_in_month_date/10e5,2)} million dollars")
print(f"Total amonut invested in this month till date is : {round(total_loan_in_month_date/10e8,2)} billion dollars")
```

```python
# from pandas.core import groupby
# good_loan=df[df['loan_status']=="Fully Paid"]


# Corrected Version

Loan_status_data=df.groupby('loan_status').agg({
                          'id': 'count',
                          'dti': 'mean',
                          'int_rate': 'mean',
                          'loan_amount': 'sum',
                          'total_payment': 'sum'
                      })


# good_loan.head(3)

Loan_status_data.rename(columns={"id":"no. of applications"},inplace=True)

display(Loan_status_data)



```

## Good Loan Analysis

```python
 # Get good loan data (Fully Paid and Current)

good_loan_data_combined=Loan_status_data.loc[['Fully Paid','Current']]

#total good loan amnt=current + fully paid
good_loan_applications = good_loan_data_combined['no. of applications'].sum()

total_applications = Loan_status_data['no. of applications'].sum()


good_loan_percentage = (good_loan_applications / total_applications) * 100

#funded amnt
good_loan_funded_amount = good_loan_data_combined['loan_amount'].sum()

# Good loan total received amount
good_loan_total_received = good_loan_data_combined['total_payment'].sum()


print(f"Good loan applications: { round(good_loan_applications,2)}")
print(f"Good loan application percentage: {round(good_loan_percentage,2)}%")
print(f"Good loan funded amount: ${round(good_loan_funded_amount,2)}")
print(f"Good loan total received amount: ${round(good_loan_total_received,2)}")
```

# **BAD loan Analysis**

```python
bad_loan_status=Loan_status_data.loc['Charged Off']

bad_loan_app=bad_loan_status['no. of applications']


total_loan_appp=Loan_status_data['no. of applications'].sum()

bad_loan_percentage=(bad_loan_app/total_loan_appp)*100

## funded amnt:

total_funded_amnt=bad_loan_status['loan_amount']

total_payment_recieved=bad_loan_status['total_payment']


print(f"Bad loan applications: {bad_loan_app}")
print(f"Bad loan application percentage: {round(bad_loan_percentage,2)}%")
print(f"Bad loan funded amount: ${round(total_funded_amnt,2)}")
print(f"Bad loan total received amount: ${round(total_payment_recieved,2)}")

```
